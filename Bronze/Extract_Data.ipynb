```python
%sql

use dw01.bronze;

create or replace temporary view v_Customers 
using csv 
options (
  path = "/Volumes/dw01/bronze/basededatos/customers.csv",
  header = "true",
  delimiter = ";"
);

create or replace table t_Customers as 
SELECT `Customer Name` as Customer_Name,Address,City,State,`Customer Type` as Customer_Type FROM v_Customers;

SELECT * from t_Customers
LIMIT 5;
```
