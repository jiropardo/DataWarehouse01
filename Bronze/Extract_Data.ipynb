```python
%sql

use dw01.bronze;

create or replace temporary view v_Customers 
using csv 
options (
  path = "/Volumes/dw01/bronze/basededatos/customers.csv",
  header = "true",
  delimiter = ";"
);

create table if not exists t_Customers as 
SELECT `Customer Name` as Customer_Name,Address as Street ,City,State,`Customer Type` as Customer_Type, `Account Manager` as Account_Manager FROM v_Customers;

SELECT * from t_Customers
LIMIT 5;
```

```python
%sql

use dw01.bronze;

create or replace temporary view v_Orders 
using csv 
options (
  path = "/Volumes/dw01/bronze/basededatos/data_proyecto.csv",
  header = "true",
  delimiter = ";"
);

create table if not exists t_Orders as 
select `Order No` as Order_No,`Order Date` as Order_Date,`Customer Name` as Customer_Name,`Address` as Address, City,`State` as State,`Customer Type` as Customer_Type,`Account Manager` as Account_Manager,`Order Priority` as Order_Priority,`Product Name` as Product_Name,`Product Category` as Product_Category,`Product Container` as Product_Container,`Ship Mode` as Ship_Mode,`Ship Date` as Ship_Date,`Cost Price` as Cost_Price,`Retail Price` as Retail_Price,`Profit Margin` as Profit_Margin,`Order Quantity` as Order_Quantity,`Sub Total` as Sub_Total,`Discount %` as Discount_Percentage,`Discount $` as Discount_Amount,`Order Total` as Order_Total,`Shipping Cost` as Shipping_Cost, Total from v_Orders;

select * from t_Orders
limit 5;
```
