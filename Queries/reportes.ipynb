```python
%sql
USE dw01.gold;

SELECT 
    c.state,
    SUM(o.order_total) AS total_sales,
    SUM(o.profit_margin) AS total_profit,
    COUNT(DISTINCT o.order_id) AS total_orders,
    COUNT(DISTINCT c.customer_id) AS total_customers
FROM dw01.gold.t_orders AS o
INNER JOIN dw01.gold.t_customers AS c
    ON o.customer_id = c.customer_id
GROUP BY 
    c.state
ORDER BY 
    total_sales DESC;
```

```python
%sql
USE dw01.gold;

SELECT 
    s.Ship_Mode,
    SUM(o.order_total) AS total_sales,
    SUM(o.profit_margin) AS total_profit,
    COUNT(DISTINCT o.order_id) AS total_orders
FROM dw01.gold.t_orders AS o
LEFT JOIN dw01.gold.t_shippingdetails AS s
    ON o.shipping_id = s.shipping_id
GROUP BY s.Ship_Mode
ORDER BY total_sales DESC;

```

```python
%sql
-- ordenes totales product año mes
SELECT 
    p.Product_Category,
    od.year,
    od.month,
    SUM(o.order_total) AS total_sales,
    SUM(o.profit_margin) AS total_profit,
    COUNT(DISTINCT o.order_id) AS total_orders
FROM dw01.gold.t_orders AS o
INNER JOIN dw01.gold.t_products AS p
    ON o.product_id = p.product_id
LEFT JOIN dw01.gold.t_dates AS od
    ON o.order_date_id = od.date_id
GROUP BY p.Product_Category, od.year, od.month
ORDER BY od.year, od.month, total_sales DESC;
```

```python
%sql

-- ordenes estado, customer_type, customer_category, product_container, order_year, order_month
USE dw01.gold;

SELECT
    -- Customer dimension
    c.state,
    c.customer_type,
    
    -- Product dimension
    p.Product_Category,
    p.Product_Container,
    
    -- Date dimension 
    od.year AS order_year,
    od.month AS order_month,
    
    -- 
    SUM(o.order_total) AS total_sales,
    SUM(o.profit_margin) AS total_profit,
    SUM(o.order_quantity) AS total_quantity,
    COUNT(DISTINCT o.order_id) AS total_orders,
    COUNT(DISTINCT o.customer_id) AS total_customers,
    AVG(o.order_total) AS avg_order_value
FROM dw01.gold.t_orders AS o
INNER JOIN dw01.gold.t_customers AS c
    ON o.customer_id = c.customer_id
INNER JOIN dw01.gold.t_products AS p
    ON o.product_id = p.product_id
LEFT JOIN dw01.gold.t_shippingdetails AS s
    ON o.shipping_id = s.shipping_id
LEFT JOIN dw01.gold.t_dates AS od
    ON o.order_date_id = od.date_id
GROUP BY 
    c.state,
    c.customer_type,
    p.Product_Category,
    p.Product_Container,
    od.year,
    od.month
ORDER BY 
    order_year, order_month, total_sales DESC;
```
